```python
from collections import Counter

def summarize_input_dir(base_path):
    """
    返回输入目录的文件统计信息。
    - 文件总数
    - 各文件类型数量（按后缀）
    - 子文件夹数量
    - 子文件夹名称列表
    """
    root = Path(base_path)
    files = [p for p in root.rglob("*") if p.is_file()]
    total_files = len(files)
    type_counts = Counter(p.suffix.lstrip(".").lower() for p in files)
    dirs = [p for p in root.iterdir() if p.is_dir()]
    dir_names = [p.name for p in dirs]
    return total_files, dict(type_counts), len(dirs), dir_names
```
